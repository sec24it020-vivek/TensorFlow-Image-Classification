```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np

# 1. Define image generator and directory paths (update directories as needed)
test_datagen = ImageDataGenerator(rescale=1./255)
train_datagen = ImageDataGenerator(rescale=1./255)

# Placeholders for directories
test_dir = '/content/sample_data'  # Please replace with your actual test directory path
train_dir = '/content/sample_data' # Please replace with your actual training directory path

# 2. Dummy generator setup to prevent NameErrors
try:
    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=32,
        class_mode='categorical'
    )
except Exception as e:
    print("Note: train_generator could not be loaded because the directory might be empty. Please update train_dir. Error:", e)
    # Creating a mock/dummy generator fallback for structural execution
    train_generator = [np.zeros((32, 224, 224, 3))]

# 3. Dummy models and histories to allow the plotting/prediction tasks to run
# Replace these placeholder structures with your actual trained models and histories
class DummyHistory:
    def __init__(self):
        self.history = {
            'accuracy': [0.5, 0.6, 0.7],
            'val_accuracy': [0.55, 0.58, 0.68],
            'loss': [1.0, 0.8, 0.6],
            'val_loss': [0.95, 0.85, 0.7]
        }

history_extract = DummyHistory()
history_fine = DummyHistory()

# Creating simple sequential models as placeholders
extract_feat_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(224, 224, 3)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(2, activation='softmax')
])

fine_tune_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(224, 224, 3)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(2, activation='softmax')
])
```

```python
# Task 1: Print TensorFlow Version
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow Version:", tf.__version__)


# Task 2: Create test_generator
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)


# Task 3: Print Length of train_generator
print("Train Generator Length:", len(train_generator))


# Task 4: Print Model Summary
extract_feat_model.summary()


# Task 5: Compile the Model
extract_feat_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)


# Task 6: Plot Accuracy Curves (Extract Features Model)
plt.plot(history_extract.history['accuracy'])
plt.plot(history_extract.history['val_accuracy'])
plt.title('Extract Features Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(['Training', 'Validation'])
plt.show()


# Task 7: Plot Loss Curves (Fine-Tuned Model)
plt.plot(history_fine.history['loss'])
plt.plot(history_fine.history['val_loss'])
plt.title('Fine-Tuned Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(['Training', 'Validation'])
plt.show()


# Task 8: Plot Accuracy Curves (Fine-Tuned Model)
plt.plot(history_fine.history['accuracy'])
plt.plot(history_fine.history['val_accuracy'])
plt.title('Fine-Tuned Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(['Training', 'Validation'])
plt.show()


# Task 9: Plot Test Image (Extract Features Model)
index_to_plot = 1
images, labels = test_generator[0]
image = images[index_to_plot]

prediction = extract_feat_model.predict(
    np.expand_dims(image, axis=0),
    verbose=0
)

class_names = list(test_generator.class_indices.keys())

plt.imshow(image.astype('uint8'))
plt.title("Extract Features Prediction: " +
          class_names[np.argmax(prediction)])
plt.axis('off')
plt.show()


# Task 10: Plot Test Image (Fine-Tuned Model)
index_to_plot = 1
image = images[index_to_plot]

prediction = fine_tune_model.predict(
    np.expand_dims(image, axis=0),
    verbose=0
)

plt.imshow(image.astype('uint8'))
plt.title("Fine-Tuned Prediction: " +
          class_names[np.argmax(prediction)])
plt.axis('off')
plt.show()
```
