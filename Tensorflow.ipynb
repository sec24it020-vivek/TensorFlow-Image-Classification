```python
import numpy as np

index_to_plot = 1
test_images, test_labels = test_generator[index_to_plot]

# index_to_plot selects the second batch; [0] selects its first image.
image = test_images[0]
true_label = int(np.argmax(test_labels[0]))

prediction = extract_feat_model.predict(np.expand_dims(image, axis=0), verbose=0)
predicted_label = int(np.argmax(prediction[0]))
class_names = list(test_generator.class_indices.keys())

plt.figure(figsize=(6, 6))
plt.imshow(np.clip(image, 0, 1))
plt.title(
    f"Extract Features Model\\nActual: {class_names[true_label]}\\n"
    f"Predicted: {class_names[predicted_label]}"
)
plt.axis("off")
plt.show()
```
